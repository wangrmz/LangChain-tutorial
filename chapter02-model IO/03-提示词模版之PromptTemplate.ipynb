## 1、PromptTemplate如何获取实例

方式1:使用构造方法的方式

```python
from langchain_core.prompts import PromptTemplate

# 定义模板（变量用 {变量名} 表示）
template = "你是一名{role}工程师，请帮我设计{feature}的单元测试用例"

# 创建模板实例
# 必填参数 input_variables、template
prompt = PromptTemplate(
    input_variables=["role", "feature"],
    template=template
)

# 填充实例中的变量
# 格式化生成最终提示词
formatted_prompt = prompt.format(role="资深测试", feature="条件覆盖")
print(formatted_prompt)


# 输出：你是一名资深测试工程师，请帮我设计条件覆盖的单元测试用例
```

方式2:from_template() 推荐

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("你是一名{role}工程师，请帮我设计{feature}用例")
formatted = prompt.format(role="测试", feature="边界值")
print(formatted)  # 输出：你是一名资深测试工程师，请帮我设计条件覆盖的单元测试用例
```

如果提示词模版中不包括变量，则调用format()时，不需要传入参数

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("你是一名工程师，请帮我设计测试用例")
formatted = prompt.format()
print(formatted)  # 输出：你是一名工程师，请帮我设计测试用例
```

## 2、两种特殊结构的使用（部分提示词模版的使用、组合提示词的使用）




### 2.1 部分提示词模版的使用（重点）

举例：在PromptTemplate的构造方法或者from_template中使用,可以进行提前赋值

```python
from langchain_core.prompts import PromptTemplate

# 定义多变量模版
# partial_variables
prompt = PromptTemplate.from_template(template="请评价{product}的优缺点，包括{aspect1}和{aspect2}",
                                      partial_variables={"aspect1": "电池续航"})
# 使用模版生成提示词
formatted = prompt.format(product="智能手机", aspect2="拍照质量")

print(formatted)  # 输出：请评价智能手机的优缺点，包括电池续航和电池续航
```

```python
from langchain_core.prompts import PromptTemplate

# 定义模板（变量用 {变量名} 表示）
template = "请评价{product}的优缺点，包括{aspect1}和{aspect2}"

# 创建模板实例
# 必填参数 input_variables、template
prompt = PromptTemplate(
    input_variables=["product", "aspect1", "aspect2"],
    template=template,
    partial_variables={"aspect1": "电池续航"}
)

# 填充实例中的变量
# 格式化生成最终提示词
formatted = prompt.format(product="智能手机", aspect2="拍照质量")
print(formatted)
# 输出：请评价智能手机的优缺点，包括电池续航和拍照质量
```

方式2：调用方法partial()

```python
from langchain_core.prompts import PromptTemplate

# 定义模板（变量用 {变量名} 表示）
template = "请评价{product}的优缺点，包括{aspect1}和{aspect2}"

# 创建模板实例
# 必填参数 input_variables、template
# prompt = PromptTemplate(
#     input_variables=["product", "aspect1", "aspect2"],
#     template=template
# )
# # partial() 调用完以后，不会对调用者这个模版对象产生影响：返回值是一个新的模版
# prompt = prompt.partial(aspect1="电池续航",aspect2="拍照质量")

# 简化写法
prompt = PromptTemplate(
    input_variables=["product", "aspect1", "aspect2"],
    template=template
).partial(aspect1="电池续航", aspect2="拍照质量")

formatted = prompt.format(product="智能手机")
print(formatted)

# 输出：请评价智能手机的优缺点，包括电池续航和拍照质量
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("请评价{product}的优缺点，包括{aspect1}和{aspect2}").partial(aspect1="电池续航",
                                                                                                  aspect2="拍照质量")

formatted = prompt.format(product="智能手机")
print(formatted)

# 输出：请评价智能手机的优缺点，包括电池续航和拍照质量
```

### 2.2 组合提示词的使用（了解）

```python
from langchain_core.prompts import PromptTemplate

template = (
        PromptTemplate.from_template("Tell me a joke about {topic}")
        + ", make it funny"
        + "\n\nand in {language}"
)
prompt = template.format(topic="sports", language="spanish")
print(prompt)

'''
Tell me a joke about sports, make it funny

and in spanish
'''
```

## 3.给变量赋值的两种方式

format():参数部分：给变量赋值 返回值：str

invoke():参数部分：字典dict 返回值：StringPromptValue --推荐使用



举例：调用format()

```python
from langchain_core.prompts import PromptTemplate

# 定义多变量模版
template = PromptTemplate.from_template(template="请评价{product}的优缺点，包括{aspect1}和{aspect2}")
# 使用模版生成提示词
formatted = template.format(product="智能手机", aspect1="电池续航", aspect2="拍照质量")

print(formatted)
print(type(formatted))  #<class 'str'>
```

举例：调用invoke()

```python
from langchain_core.prompts import PromptTemplate

# 定义多变量模版
template = PromptTemplate.from_template(template="请评价{product}的优缺点，包括{aspect1}和{aspect2}")
# 使用invoke()方法，查看源码传入字典dict类型
formatted = template.invoke(input={"product": "智能手机", "aspect1": "电池续航", "aspect2": "拍照质量"})

print(formatted)  # text='请评价智能手机的优缺点，包括电池续航和拍照质量'
print(type(formatted))  # <class 'langchain_core.prompt_values.StringPromptValue'>

```

# 4、结合大模型的使用

```python
import os
import dotenv
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_openai import ChatOpenAI

# 加载配置文件
dotenv.load_dotenv()
# 设置环境
os.environ['OPENAI_API_KEY'] = os.getenv("OPENAI_API_KEY1")
os.environ['OPENAI_BASE_URL'] = os.getenv("OPENAI_BASE_URL")
# 获取对话模型：
chat_model = ChatOpenAI(model="gpt-4o-mini",max_tokens=1000)

# 生成提示词模版
template = PromptTemplate.from_template(template="请评价{product}的优缺点，包括{aspect1}和{aspect2}")
# 使用invoke()方法，查看源码传入字典dict类型
prompt = template.invoke(input={"product": "智能手机", "aspect1": "电池续航", "aspect2": "拍照质量"})

# 调用大模型，将提示词传入
response = chat_model.invoke(prompt)
print(response)
print(type(response))
```
